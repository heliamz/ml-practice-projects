```python
from transformers import pipeline
import warnings
warnings.filterwarnings('ignore')
```

## Initialize Text Generation Pipeline


```python
# Initialize the text generation pipeline
generator = pipeline('text-generation')
print(f"Pipeline initialized: {type(generator)}")
print("Using default model (typically GPT-2)")
```

## Basic Text Generation


```python
# Simple text generation example
prompt = "The future of artificial intelligence"
result = generator(prompt, max_new_tokens=50, num_return_sequences=1)

print(f"Prompt: {prompt}")
print(f"\nGenerated text:")
print(result[0]['generated_text'])
```

## Multiple Generation Examples


```python
# Various text generation prompts
prompts = [
    "Write a short story about a data scientist who discovers a hidden pattern in city traffic data.",
    "Explain quantum computing to a high school student using simple analogies.",
    "Generate a product review for a noise-canceling headphone after one week of use.",
    "Continue the following paragraph: The experiment failed, but the results were not what anyone expected.",
    "Write a dialogue between a human and an AI debating creativity.",
    "Summarize the key causes of climate change in one paragraph.",
    "Create a fictional news article about a breakthrough in renewable energy.",
    "Write an opening paragraph for a mystery novel set in a small coastal town."
]

print("Text Generation Examples:\n")
print("=" * 80)

for i, prompt in enumerate(prompts, 1):
    print(f"\nExample {i}:")
    print(f"Prompt: {prompt}")
    result = generator(prompt, max_new_tokens=100, num_return_sequences=1, 
                      do_sample=True, temperature=0.7)
    generated = result[0]['generated_text']
    print(f"Generated: {generated}")
    print("-" * 80)
```

```python
# Generate with different parameters
prompt = "The impact of machine learning on healthcare"

# Conservative generation (low temperature)
result_conservative = generator(
    prompt, 
    max_new_tokens=50, 
    num_return_sequences=1,
    temperature=0.3,  # Lower temperature = more deterministic
    do_sample=True
)

# Creative generation (high temperature)
result_creative = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1,
    temperature=0.9,  # Higher temperature = more creative/random
    do_sample=True
)

print("Prompt:", prompt)
print("\n" + "=" * 80)
print("Conservative Generation (temperature=0.3):")
print(result_conservative[0]['generated_text'])
print("\n" + "=" * 80)
print("Creative Generation (temperature=0.9):")
print(result_creative[0]['generated_text'])
```

```python
# Generate multiple variations of the same prompt
prompt = "Artificial intelligence will"
results = generator(
    prompt,
    max_new_tokens=60,
    num_return_sequences=3,  # Generate 3 different completions
    do_sample=True,
    temperature=0.7
)

print(f"Prompt: {prompt}")
print("\n" + "=" * 80)
for i, result in enumerate(results, 1):
    print(f"\nVariation {i}:")
    print(result['generated_text'])
    print("-" * 80)
```

## Results Summary

The text generation pipeline uses autoregressive language models (like GPT-2) to generate coherent text continuations.

**Key Features:**
- **Autoregressive Generation**: Generates text token by token
- **Configurable Parameters**: Control creativity, length, and diversity
- **Multiple Variations**: Generate different completions for the same prompt
- **Zero-shot**: Works without fine-tuning

**Parameters:**
- `max_new_tokens`: Maximum number of tokens to generate
- `temperature`: Controls randomness (0.0 = deterministic, 1.0+ = creative)
- `num_return_sequences`: Number of different completions to generate
- `do_sample`: Whether to use sampling (True) or greedy decoding (False)

**Use Cases:**
- Content creation and writing assistance
- Story continuation
- Code completion
- Creative writing prompts
- Text summarization (with appropriate prompts)
